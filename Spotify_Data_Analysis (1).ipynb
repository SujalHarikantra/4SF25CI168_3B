```python
df.drop_duplicates(inplace=True)
```

```python
df.drop(columns=['track_id'], inplace=True)
```

```python
df['popularity'] = pd.to_numeric(
    df['popularity'], errors='coerce'
)
```

```python
df['duration_minutes'] = df['duration_ms'] / 60000
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.isnull().sum().sum()
```

```python
for col in df.select_dtypes(include=np.number).columns:
    df[col] = df[col].fillna(df[col].mean())
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['popularity'], bins=20)
plt.title('Spotify Song Popularity')
plt.xlabel('Popularity')
plt.ylabel('Number of Songs')
plt.show()
```

```python
top_artists = df['artists'].value_counts().head(10)

top_artists.plot(kind='bar', figsize=(8,5))
plt.title('Top 10 Artists')
plt.xlabel('Artists')
plt.ylabel('Number of Songs')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df['energy'], df['danceability'])
plt.title('Energy vs Danceability')
plt.xlabel('Energy')
plt.ylabel('Danceability')
plt.show()
```

```python
sns.boxplot(x=df['popularity'])
plt.title('Song Popularity Distribution')
plt.show()
```
